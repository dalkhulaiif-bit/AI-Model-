```python
import cv2
from ultralytics import YOLO

# 1. استخدام نموذج YOLOv8 الخفيف الجاهز للاستخدام الفوري (سيتم تحميله تلقائياً عند أول تشغيل)
# هذا النموذج يتعرف على الأشياء العامة (مثل الأشخاص، الهواتف، إلخ) كبداية للتجربة والتأكد من عمل الكاميرا
model = YOLO("yolov8n.pt")

# 2. فتح الكاميرا المباشرة عبر OpenCV
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open webcam.")
    exit()

print("--- Starting YOLO Live Stream --- Press 'q' to exit ---")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame.")
        break

    # 3. تطبيق النموذج على إطار الكاميرا
    results = model(frame)

    # 4. رسم المربعات والتسميات تلقائياً على الإطار
    annotated_frame = results[0].plot()

    # 5. عرض النتيجة مباشرة
    cv2.imshow("YOLO Live Stream", annotated_frame)

    # الضغط على زر 'q' لإغلاق البث
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# تنظيف وإغلاق الكاميرا
cap.release()
cv2.destroyAllWindows()
print("Live stream closed.")
```
